### 2. Définition de l'architecture du modèle

Nous allons utiliser un modèle séquentiel Keras avec des couches:
-   `Embedding`: Pour convertir chaque caractère numérique en un vecteur dense.
-   `GRU` (Gated Recurrent Unit): Un type de RNN, efficace pour le traitement des séquences.
-   `Dense`: La couche de sortie pour prédire le prochain caractère.

```python
# @title
# Paramètres du modèle
vocab_size = len(vocab)
embedding_dim = 256
GRU_units = 1024

# Construction du modèle
def build_model(vocab_size, embedding_dim, GRU_units, batch_size):
    model = Sequential([
        tf.keras.layers.Input(batch_shape=(batch_size, None)), # Define input shape using an Input layer
        Embedding(vocab_size, embedding_dim),
        GRU(GRU_units, return_sequences=True, stateful=True, recurrent_initializer='glorot_uniform'),
        Dense(vocab_size)
    ])
    return model

model = build_model(vocab_size, embedding_dim, GRU_units, BATCH_SIZE)
model.summary()
```

### 3. Entraînement du modèle

Nous allons compiler le modèle avec un optimiseur et une fonction de perte appropriés, puis l'entraîner sur nos données.

```python
# @title
# Fonction de perte personnalisée
def loss(labels, logits):
    return sparse_categorical_crossentropy(labels, logits, from_logits=True)

# Compiler le modèle
model.compile(optimizer=Adam(), loss=loss)

# Nombre d'époques pour l'entraînement (peut être ajusté)
EPOCHS = 50

# Entraîner le modèle
history = model.fit(dataset, epochs=EPOCHS)

# Sauvegarder les poids du modèle si vous le souhaitez
# model.save_weights('./my_text_generator_weights.h5')
```

### 4. Génération de texte

Après l'entraînement, nous pouvons utiliser le modèle pour générer de nouvelles séquences de texte. Nous devrons recharger le modèle avec une taille de batch de 1 pour pouvoir générer un caractère à la fois.

```python
# @title
# Reconstruire le modèle avec une taille de batch de 1 pour la génération
model_for_generation = build_model(vocab_size, embedding_dim, GRU_units, batch_size=1)

# Charger les poids du modèle entraîné
model_for_generation.set_weights(model.get_weights())
model_for_generation.build(tf.TensorShape([1, None]))

def generate_text(model, start_string, num_generate=100):
    # Nombre de caractères à générer
    # Convertir la chaîne de départ en nombres (en utilisant notre mapping char_to_int)
    input_eval = [char_to_int[s] for s in start_string]
    input_eval = tf.expand_dims(input_eval, 0)

    # Liste pour stocker les résultats des caractères générés
    text_generated = []

    # Température pour le mode échantillonnage: valeurs plus élevées = plus de créativité
    temperature = 1.0

    # Réinitialiser l'état du RNN (important pour les modèles stateful)
    model.layers[1].reset_states() # Correction: Appeler reset_states sur la couche GRU

    for i in range(num_generate):
        predictions = model(input_eval)
        # Supprimer la dimension du batch
        predictions = tf.squeeze(predictions, 0)

        # Utiliser une distribution catégorique pour échantillonner le caractère prédit
        predictions = predictions / temperature
        predicted_id = tf.random.categorical(predictions, num_samples=1)[-1,0].numpy()

        # Passer le caractère prédit comme entrée suivante au modèle
        input_eval = tf.expand_dims([predicted_id], 0)

        text_generated.append(int_to_char[predicted_id])

    return start_string + ''.join(text_generated)

# Générer du texte en commençant par une phrase donnée
print(generate_text(model_for_generation, start_string="quand les poules auront des "))
```
