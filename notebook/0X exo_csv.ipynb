```python
# ============================================
# 1. IMPORT ET CHARGEMENT DU FICHIER CSV
# ============================================

from google.colab import files
import pandas as pd

# Télécharger le fichier CSV
uploaded = files.upload()

file_name = None
if uploaded:
    for fn in uploaded.keys():
        print(f'Fichier "{fn}" téléchargé.')
        file_name = fn
else:
    print("Aucun fichier n'a été téléchargé. Veuillez le télécharger manuellement ou exécuter cette cellule à nouveau.")

# Charger le fichier CSV dans un DataFrame pandas
df = pd.read_csv('exo.csv')
print(f"DataFrame chargé avec succès ! Dimensions: {df.shape}")
```

```python
# ============================================
# 2. NETTOYAGE DES DONNÉES - COLONNE PRIX
# ============================================

# Convertir la colonne prix_dollars en string, supprimer le symbole '$', puis convertir en float
df['prix_dollars'] = df['prix_dollars'].astype(str).apply(lambda x: x.replace('$', ''))
df['prix_dollars'] = df['prix_dollars'].astype(float)

print("✓ Symboles '$' supprimés et données converties en float")
print(f"Aperçu des prix:\n{df['prix_dollars'].head()}")
```

```python
# Calculer la moyenne des prix (arrondie à 2 décimales)
# Cette valeur sera utilisée pour remplacer les prix manquants
moyenne_prix = df['prix_dollars'].mean().round(2)
print(f"Moyenne des prix: ${moyenne_prix}")
```

```python
# Remplacer les valeurs manquantes (NaN) par la moyenne calculée
df['prix_dollars'].fillna(moyenne_prix, inplace=True)
print(f"✓ Valeurs manquantes remplies avec la moyenne: ${moyenne_prix}")
print(f"Nombre de valeurs manquantes restantes: {df['prix_dollars'].isna().sum()}")
```

```python
# Afficher le DataFrame après nettoyage
print("DataFrame après nettoyage des prix:")
df
```

```python
# ============================================
# 3. ENRICHISSEMENT DES DONNÉES - CODES PAYS
# ============================================

# Créer un dictionnaire de correspondance pays -> codes ISO
abreviation = {'France': 'FR', 'Canada': 'CA', 'Belgium': 'BE'}

# Ajouter une nouvelle colonne avec les codes pays
df['codes_pays'] = df['pays'].map(abreviation)
print("✓ Colonne 'codes_pays' ajoutée")
print(f"Codes pays uniques: {df['codes_pays'].unique()}")
```

```python
# Afficher le DataFrame avec la nouvelle colonne
print("DataFrame avec codes pays:")
df
```

```python
# ============================================
# 4. ANALYSE DES DONNÉES
# ============================================

# Calculer la somme des prix par genre
prix_par_genre = df.groupby('genre')['prix_dollars'].sum()
print("Somme des prix par genre:")
prix_par_genre
```

```python
# ============================================
# 5. CONVERSION EN EUROS
# ============================================

# Convertir les prix de dollars en euros (taux: 1$ = 0.92€)
df['prix_euros'] = (df['prix_dollars'] * 0.92).round(0)
print("✓ Colonne 'prix_euros' ajoutée (taux: 1$ = 0.92€)")
```
