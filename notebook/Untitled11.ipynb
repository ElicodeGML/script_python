### Transformer un DataFrame Pandas en tableau NumPy

```python
import pandas as pd
import numpy as np

# Créeons un exemple de DataFrame
data = {'col1': [1, 2, 3], 'col2': [4, 5, 6]}
df = pd.DataFrame(data)

print("Voici votre DataFrame:\n")
display(df)
```

Pour convertir ce DataFrame en un tableau NumPy, vous pouvez utiliser la méthode `.to_numpy()` :

```python
# Conversion du DataFrame en tableau NumPy
numpy_array = df.to_numpy()

print("Voici le tableau NumPy résultant:\n")
display(numpy_array)

print(f"Le type de la variable est maintenant : {type(numpy_array)}")
```

### Enlever la dernière colonne d'un tableau NumPy

```python
import numpy as np

# Créeons un exemple de tableau NumPy avec plusieurs colonnes
sample_array = np.array([
    [10, 20, 30, 40],
    [11, 21, 31, 41],
    [12, 22, 32, 42]
])

print("Voici le tableau NumPy original :\n")
display(sample_array)
```

Pour supprimer la dernière colonne, vous pouvez utiliser le slicing en spécifiant `[:, :-1]`. Cela sélectionne toutes les lignes (`:` ) et toutes les colonnes sauf la dernière (`:-1` ) :

```python
# Enlever la dernière colonne
array_without_last_column = sample_array[:, :-1]

print("Voici le tableau NumPy sans la dernière colonne :\n")
display(array_without_last_column)
```

Vous pouvez également utiliser `np.delete()` si vous préférez, en spécifiant l'indice de la colonne à supprimer et l'axe : `np.delete(sample_array, obj=-1, axis=1)`.
