```python
import numpy as np
from scipy import fftpack
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
x = np.linspace(0, 30, 1000)
y = 3*np.sin(x) + 2*np.sin(5*x) + np.sin(10*x) + 5 * np.random.randn(x.size)
plt.plot(x, y)
```

```python
from scipy import fftpack
```

```python
fourrier = fftpack.fft(y)
frequences = fftpack.fftfreq(y.size)
power = np.abs(fourrier)
plt.plot(np.abs(frequences), power)
```

```python
fourrier[power < 400] = 0
power = np.abs(fourrier)
plt.plot(np.abs(frequences), power)
```

```python
Q = fftpack.ifft(fourrier)
```

```python
plt.figure()
plt.plot(x, y, lw=0.5)
plt.plot(x, Q, lw=3)
```
