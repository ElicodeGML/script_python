```python
import math

class Calculatrice:
    """Classe pour des opérations arithmétiques et des fonctions de base."""

    def add(self, a, b):
        """Retourne la somme de a et b."""
        return a + b

    def subtract(self, a, b):
        """Retourne la différence de a et b."""
        return a - b

    def multiply(self, a, b):
        """Retourne le produit de a et b."""
        return a * b

    def divide(self, a, b):
        """Retourne le quotient de a et b. Gère la division par zéro."""
        if b == 0:
            raise ValueError("Cannot divide by zero!")
        return a / b

    def power(self, base, exponent):
        """Retourne base élevé à la puissance exponent."""
        return base ** exponent

    def square_root(self, x):
        """Retourne la racine carrée de x."""
        if x < 0:
            raise ValueError("Cannot calculate square root of a negative number.")
        return math.sqrt(x)

# Démonstration de la classe Calculatrice
calc = Calculatrice()
print("--- Démonstration Calculatrice ---")
print(f"5 + 3 = {calc.add(5, 3)}")
print(f"10 - 4 = {calc.subtract(10, 4)}")
print(f"6 * 7 = {calc.multiply(6, 7)}")
print(f"20 / 5 = {calc.divide(20, 5)}")
print(f"2^3 = {calc.power(2, 3)}")
print(f"Racine carrée de 16 = {calc.square_root(16)}")
```

```python
# @title
import statistics
import math
from collections import Counter

class Statistiques:
    """Classe pour calculer diverses mesures statistiques sur une liste de nombres."""

    def mean(self, data_list):
        """Retourne la moyenne d'une liste de nombres."""
        if not data_list: raise ValueError("List cannot be empty.")
        return sum(data_list) / len(data_list)

    def median(self, data_list):
        """Retourne la médiane."""
        if not data_list: raise ValueError("List cannot be empty.")
        sorted_list = sorted(data_list)
        n = len(sorted_list)
        if n % 2 == 0:
            # Si le nombre d'éléments est pair
            mid1 = sorted_list[n // 2 - 1]
            mid2 = sorted_list[n // 2]
            return (mid1 + mid2) / 2
        else:
            # Si le nombre d'éléments est impair
            return sorted_list[n // 2]

    def mode(self, data_list):
        """Retourne le(s) mode(s). Peut retourner une liste si plusieurs modes."""
        if not data_list: raise ValueError("List cannot be empty.")
        counts = Counter(data_list)
        max_count = max(counts.values())
        modes = [key for key, value in counts.items() if value == max_count]
        return modes

    def std_deviation(self, data_list, sample=True):
        """Retourne l'écart-type. 'sample=True' pour l'écart-type d'échantillon, 'False' pour la population."""
        if len(data_list) < 2 and sample: raise ValueError("Need at least two data points for sample standard deviation.")
        if not data_list: raise ValueError("List cannot be empty.")
        return statistics.stdev(data_list) if sample else statistics.pstdev(data_list)

    def variance(self, data_list, sample=True):
        """Retourne la variance. 'sample=True' pour la variance d'échantillon, 'False' pour la population."""
        if len(data_list) < 2 and sample: raise ValueError("Need at least two data points for sample variance.")
        if not data_list: raise ValueError("List cannot be empty.")
        return statistics.variance(data_list) if sample else statistics.pvariance(data_list)

    def data_range(self, data_list):
        """Retourne l'étendue (max - min) d'une liste de nombres."""
        if not data_list: raise ValueError("List cannot be empty.")
        return max(data_list) - min(data_list)

# Démonstration de la classe Statistiques
stats = Statistiques()
data = []
import random
x = random.randint(10, 20)
for i in range(x):
  y = random.randint(0, 10)
  data.append(y)
print("classe STAT")
print(f"Données: {data}")
print(f"Moyenne: {stats.mean(data)}")
print(f"Médiane: {stats.median(data)}")
print(f"Écart-type (échantillon): {stats.std_deviation(data)}")
print(f"Variance (échantillon): {stats.variance(data)}")
print(f"Dif. min./max. :{stats.data_range(data)}")
```

```python
# @title
import math

class Géométrie:
    """Classe pour des calculs liés aux formes géométriques."""

    def circle_area(self, radius):
        """Calcule l'aire d'un cercle."""
        if radius < 0: raise ValueError("Radius cannot be negative.")
        return math.pi * (radius ** 2)

    def rectangle_area(self, length, width):
        """Calcule l'aire d'un rectangle."""
        if length < 0 or width < 0: raise ValueError("Length and width cannot be negative.")
        return length * width

    def triangle_area(self, base, height):
        """Calcule l'aire d'un triangle."""
        if base < 0 or height < 0: raise ValueError("Base and height cannot be negative.")
        return 0.5 * base * height

    def hypotenuse(self, side1, side2):
        """Calcule l'hypoténuse d'un triangle rectangle (Théorème de Pythagore)."""
        if side1 < 0 or side2 < 0: raise ValueError("Sides cannot be negative.")
        return math.sqrt(side1**2 + side2**2)

    def distance_between_points(self, x1, y1, x2, y2):
        """Calcule la distance entre deux points 2D."""
        return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# Démonstration de la classe Géométrie
geo = Géométrie()
print("\n--- Démonstration Géométrie ---")
print(f"Aire d'un cercle de rayon 5: {geo.circle_area(5):.2f}")
print(f"Aire d'un rectangle de longueur 4 et largeur 6: {geo.rectangle_area(4, 6)}")
print(f"Aire d'un triangle de base 10 et hauteur 5: {geo.triangle_area(10, 5)}")
print(f"Hypoténuse d'un triangle avec côtés 3 et 4: {geo.hypotenuse(3, 4)}")
print(f"Distance entre (0,0) et (3,4): {geo.distance_between_points(0, 0, 3, 4)}")
```

```python
# @title
class Series:
    """Classe pour générer et manipuler des suites numériques."""

    def fibonacci(self, n):
        """Génère la suite de Fibonacci jusqu'au n-ième terme."""
        if n < 0: raise ValueError("n cannot be negative.")
        if n == 0: return []
        if n == 1: return [0]

        fib_series = [0, 1]
        while len(fib_series) < n:
            fib_series.append(fib_series[-1] + fib_series[-2])
        return fib_series

    def arithmetic_sum(self, start, count, common_diff):
        """Calcule la somme d'une série arithmétique."""
        if count < 0: raise ValueError("Count cannot be negative.")
        if count == 0: return 0

        # Formule de la somme d'une série arithmétique: n/2 * (2*a + (n-1)*d)
        return (count / 2) * (2 * start + (count - 1) * common_diff)

    def geometric_sum(self, start, count, common_ratio):
        """Calcule la somme d'une série géométrique."""
        if count < 0: raise ValueError("Count cannot be negative.")
        if count == 0: return 0

        if common_ratio == 1: # Cas spécial pour ratio = 1
            return start * count

        # Formule de la somme d'une série géométrique: a * (1 - r^n) / (1 - r)
        return start * (1 - common_ratio**count) / (1 - common_ratio)

    def prime_numbers_up_to(self, n):
        """Génère tous les nombres premiers jusqu'à n (crible d'Ératosthène)."""
        if n < 2: return []

        is_prime = [True] * (n + 1)
        is_prime[0] = is_prime[1] = False

        for p in range(2, int(n**0.5) + 1):
            if is_prime[p]:
                for multiple in range(p*p, n + 1, p):
                    is_prime[multiple] = False

        primes = [p for p, prime in enumerate(is_prime) if prime]
        return primes

# Démonstration de la classe Series
series_gen = Series()
print("\n--- Démonstration Séries Mathématiques ---")
print(f"Fibonacci jusqu'au 8ème terme: {series_gen.fibonacci(8)}")
print(f"Somme arithmétique (début=1, nb=5, diff=2): {series_gen.arithmetic_sum(1, 5, 2)}") # 1+3+5+7+9 = 25
print(f"Somme géométrique (début=1, nb=4, ratio=2): {series_gen.geometric_sum(1, 4, 2)}") # 1+2+4+8 = 15
print(f"Nombres premiers jusqu'à 20: {series_gen.prime_numbers_up_to(20)}")
```
