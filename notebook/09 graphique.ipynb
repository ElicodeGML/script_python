```python
df
```

```python
# graphique du prix en fonction de la date
df.groupby('date')['price_paid'].sum().plot(figsize=[15, 6])
```

```python
# diagramme camembert du prix en fonction du genre
df.groupby('gender')['price_paid'].sum().plot.pie(legend=True, figsize=[5, 5])
```

```python
# diagramme barre du prix en fonction du pays
df.groupby('country')['price_paid'].sum().plot.bar(rot=0, figsize=[5, 5])
```
