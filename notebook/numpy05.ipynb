````python
# @title
import matplotlib.pyplot as plt
import seaborn as sns

# Generate a large number of samples from a standard normal distribution
samples = np.random.randn(100000)

# Create a histogram to visualize the distribution
plt.figure(figsize=(10, 6))
sns.histplot(samples, bins=50, kde=True, color='skyblue')
plt.title('Distribution des nombres générés par np.random.randn')
plt.xlabel('Valeur')
plt.ylabel('Fréquence')
plt.grid(axis='y', alpha=0.75)
plt.show()
````
