```python
df["tax"].fillna(0, inplace=True)
```

```python
df
```

```python
df.price_paid = df.price_paid.apply(lambda x: x.replace('$', ''))
df.price_paid = df.price_paid.astype(float)
```

```python
df
```

```python
# df["price_total"] = df["price_paid"] * (100 - df["price_paid"]) / 100 OU
df["price_total"] = df["price_paid"] * (1 - df["tax"] / 100)
```

```python
df
```

```python
counties = {"United States": "UN", "France":"FR", "Canada": "CA", "Morocco": "MA"}
df["country_code"] = df['country'].map(counties)
```
