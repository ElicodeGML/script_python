# <center>**ZOOMER**</center>

```python
# @title
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
astronaut_image = data.astronaut()
X = astronaut_image

plt.imshow(astronaut_image)
plt.title('Astronaut Image from scikit-image')
plt.axis('off') # Hide axes for cleaner image display
plt.show()
X.shape
```

This code zooms into the original image `X`. It takes a part of the image from row 50 to 249 (200 pixels tall) and from column 150 to 349 (200 pixels wide). This creates a new image called `zoome_X`.

```python
zoome_X = X[50:250, 150:350, :]
plt.imshow(zoome_X)
```

This code takes the `zoome_X` image and zooms in even more. It selects a part of `zoome_X` from row 0 to 124 (125 pixels tall) and from column 0 to 124 (125 pixels wide). This makes an even more zoomed-in image called `zomm_X`.

```python
zomm_X = zoome_X[0:125, 0:125, :]
zomm_X
```
