# Introduction to CyberPi, mBot2, and mbuild

The **CyberPi** is the main control board, a powerful microcontroller with a color screen, built-in sensors (accelerometer, gyroscope), buttons, Wi-Fi, Bluetooth, speaker, and microphone. It's the brain of your mBot2.

The **mBot2** is a mobile robot platform that connects to the CyberPi. It extends the CyberPi's capabilities by adding motors (with encoders), a Quad RGB sensor for line following, an ultrasonic sensor, and ports for additional `mbuild` modules.

**`mbuild`** refers to a wide ecosystem of add-on modules (sensors, actuators, displays) that can be easily connected to the CyberPi or mBot2 to expand their functionalities.

Let's break down how to use these components with Python, along with the standard `time` and event handling concepts.

## 4. The `time` Module: Delays and Timing

The standard Python `time` module is crucial for controlling the flow of your robotics programs, especially for creating delays (pauses) and measuring durations.

```python
import time
import cyberpi

print("Time Module Demo")

cyberpi.display.clear()
cyberpi.display.show_label("Start...", 0, 0)
time.sleep(1) # Pause for 1 second

cyberpi.display.clear()
cyberpi.display.show_label("Wait 2s...", 0, 0)
time.sleep(2) # Pause for 2 seconds

cyberpi.display.clear()
cyberpi.display.show_label("Done!", 0, 0)

# Measuring elapsed time
start_time = time.time()
cyberpi.display.show_label("Counting...", 0, 2)
while time.time() - start_time < 3: # Loop for 3 seconds
    elapsed = time.time() - start_time
    cyberpi.display.show_label(str(round(elapsed, 1)) + "s", 0, 3)
    time.sleep(0.1)

time.sleep(1)
cyberpi.display.clear()
print("Time Module Demo End")
```

## 5. Event Handling Concepts (Implicit in CyberPi)

While there isn't a standalone `event` module in the same way as `time` or `mbuild`, event handling is built into the `CyberPi` library through functions that detect specific occurrences, like button presses or sensor thresholds being met. You typically poll for these events within a `while True` loop.

```python
import cyberpi
import time
import random

print("Event Handling Demo")

cyberpi.display.clear()
cyberpi.display.show_label("Press A to start", 0, 0)
cyberpi.display.show_label("Press B to stop ", 0, 2)

program_running = False

while True:
    if button_a.is_pressed():
        if not program_running:
            program_running = True
            cyberpi.display.clear()
            cyberpi.display.show_label("Program Started!", 0, 0)
            cyberpi.display.show_label("(Press B to stop)", 0, 1)
            print("Program started.")
        time.sleep(0.2) # Debounce button

    if button_b.is_pressed():
        if program_running:
            program_running = False
            cyberpi.display.clear()
            cyberpi.display.show_label("Program Stopped!", 0, 0)
            cyberpi.display.show_label("(Press A to start)", 0, 1)
            print("Program stopped.")
        time.sleep(0.2) # Debounce button

    if program_running:
        # Your main robot logic goes here when the program is 'running'
        cyberpi.display.set_color(random.randint(0,255), random.randint(0,255), random.randint(0,255))
        cyberpi.display.show_label("Running...", 0, 4)
        # Example: move mBot2 forward slightly if running
        # mbot2.forward(20, 0.1)
        # mbot2.stop()

    time.sleep(0.1) # Small delay to prevent busy-waiting

    # To exit this Colab cell manually, you'd typically interrupt the kernel.
    # On the physical CyberPi, you might have another button combination or power cycle.
    # For demonstration in Colab, this loop runs indefinitely until manually stopped.

print("Event Handling Demo End (if loop was exited)")
```
