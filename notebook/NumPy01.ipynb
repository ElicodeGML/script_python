```python
import numpy as np
```

**<center>Initialisation</center>**

```python
# a table with 4 rows and 5 columns filled with 3
A = np.full((4, 5), 3)
A
```

```python
# a table with 4 rows and 5 columns filled with 1
B = np.ones((4, 5))
B
```

```python
# a table with 4 rows and 5 columns filled with 0
C = np.zeros((4, 5))
C
```

**<center>plages</center>**

```python
# a table with numbers from 0 to 10 with a step of 2
X = np.arange(0, 10, 2)
X
```

```python
# a table with 5 numbers between 0 and 10
Y = np.linspace(0, 10, 5)
Y
```

**<center>array manipulation</center>**

```python
# put A and B together
D = np.concatenate((A, B), axis=0)
D
```

```python
# put A and B together the other way
E = np.concatenate((A, B), axis=1)
E
```

```python
# make E flat into 1 line
F = E.ravel()
F
```

```python
# change F's size
G = F.reshape(8, 5)
G
```
