import l'image et les modules

```python
# @title
# Exemple pour afficher une image téléchargée
from IPython.display import Image, display

# Remplacez 'votre_image.jpg' par le nom de fichier de votre image téléchargée
image_path = 'Ane.png'

try:
    display(Image(filename=image_path))
    print(f"Image '{image_path}' affichée avec succès.")
except FileNotFoundError:
    print(f"Erreur: Le fichier '{image_path}' n'a pas été trouvé. Assurez-vous de l'avoir téléchargé dans Colab.")
except Exception as e:
    print(f"Une erreur est survenue lors de l'affichage de l'image: {e}")
```

```python
# @title
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
image_pil = Image.open(image_path)
image_pil_grid = image_pil.convert('L')
```

```python
# @title
image = np.array(image_pil_grid)
image.shape
```

voici l'image

```python
# @title
plt.imshow(image, cmap='gray')
plt.grid(True)
plt.show()
```

voici l'image avec des contraste

```python
# @title
# Réinitialisez image_zoomer en tant que tableau NumPy
image_zoomer = image[200:830, 200:570]

# rendre les pixels gris foncé noire
image_zoomer_copy = image_zoomer.copy()
image_zoomer_copy[image_zoomer_copy < 50] = 0


# rendre les pixels gris claire blanc
image_foncé = image_zoomer_copy.copy()
image_foncé[image_foncé > 175] = 255


image_fini = image_foncé.copy()

plt.imshow(image_fini, cmap='gray')
plt.grid(False)
plt.show()
```

voici l'image éclaircie

```python
# @title
image_clair = image_zoomer.copy()
image_clair_fini = image_clair + 51
plt.imshow(image_clair_fini, cmap='gray')
plt.grid(False)
plt.show()
```
