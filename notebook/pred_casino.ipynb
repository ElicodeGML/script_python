```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
data_casino = {
      'mise': [500, 20, 10000, 50, 2000, 100, 15000, 300, 50, 8000],
      'nb_parties': [5, 10, 2, 15, 8, 20, 1, 12, 10, 3],
      'reflexion': [10, 30, 2, 45, 15, 25, 1, 35, 40, 5],
      'suspicion': [10, 2, 95, 5, 40, 8, 99, 15, 4, 88] # La cible à prédire
      }
df = pd.DataFrame(data_casino)
```

```python
df['mise_par_partie'] = df['mise'] / df['nb_parties']
df['reflexion_par_partie'] = df['reflexion'] / df['nb_parties']

# Correcting x to exclude the target variable 'suspicion'
x = df[['mise', 'nb_parties', 'reflexion', 'mise_par_partie', 'reflexion_par_partie']]
y = df['suspicion']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# On crée l'usine : Scaler + IA
model_pro = make_pipeline(
        StandardScaler(),
        RandomForestRegressor(n_estimators=100, random_state=42)
        )

# Entraînement
print(f"Features used for training: {x_train.columns.tolist()}")
model_pro.fit(x_train, y_train)

# Prédiction et Score
prediction = model_pro.predict(x_test)
erreur = mean_absolute_error(y_test, prediction)
```

```python
print(erreur)
```

```python
# Original input values for mise, nb_parties, reflexion
new_mise = 10000
new_nb_parties = 25
new_reflexion = 50

# Calculate derived features for the new input, matching the training data preparation
new_mise_par_partie = new_mise / new_nb_parties
new_reflexion_par_partie = new_reflexion / new_nb_parties

# Create a DataFrame for prediction, ensuring it's 2D and has the correct feature names and order
# The order of columns must match the features used during model training.
new_data_for_prediction = pd.DataFrame([[new_mise,
                                         new_nb_parties,
                                         new_reflexion,
                                         new_mise_par_partie,
                                         new_reflexion_par_partie]],
                                       columns=['mise', 'nb_parties', 'reflexion', 'mise_par_partie', 'reflexion_par_partie'])

pred = model_pro.predict(new_data_for_prediction)
print(pred)
```
